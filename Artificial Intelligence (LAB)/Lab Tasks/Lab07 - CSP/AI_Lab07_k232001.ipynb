Q1

```python
from ortools.sat.python import cp_model

def f():
    m=cp_model.CpModel()

    a=m.NewIntVar(1,6,'a')
    b=m.NewIntVar(1,6,'b')
    c=m.NewIntVar(1,6,'c')
    d=m.NewIntVar(1,6,'d')
    e=m.NewIntVar(1,6,'e')
    g=m.NewIntVar(1,6,'g')

    m.AddAllDifferent([a,b,c,d,e,g])

    x=m.NewIntVar(-5,5,'x')
    y=m.NewIntVar(-5,5,'y')

    m.Add(x==b-d)
    m.AddAbsEquality(1,x)

    m.Add(y==a-c)
    m.AddAbsEquality(m.NewIntVar(2,5,'z'),y)

    m.AddAllowedAssignments([e],[[2],[4],[6]])
    m.Add(g!=1)

    s=cp_model.CpSolver()
    if s.Solve(m)==4:
        print(s.Value(a),s.Value(b),s.Value(c),s.Value(d),s.Value(e),s.Value(g))
```

```python
f()
```

```python
from ortools.sat.python import cp_model

def solve_course_timetabling():
    m=cp_model.CpModel()

    t=[m.NewIntVar(1,4,str(i)) for i in range(6)]
    r=[m.NewIntVar(1,3,str(i+10)) for i in range(6)]

    n=[35,50,45,60,30,80]

    m.Add(t[0]!=t[1])
    m.Add(t[1]!=t[2])
    m.Add(t[2]!=t[5])
    m.Add(t[3]!=t[4])

    for i in range(6):
        if n[i]>60:
            m.Add(r[i]==3)
        elif n[i]>40:
            m.Add(r[i]!=1)

    for i in range(6):
        for j in range(i+1,6):
            u=m.NewBoolVar(str(i)+str(j))
            m.Add(t[i]==t[j]).OnlyEnforceIf(u)
            m.Add(t[i]!=t[j]).OnlyEnforceIf(u.Not())
            m.Add(r[i]!=r[j]).OnlyEnforceIf(u)

    m.AddAllowedAssignments([t[5]],[[3],[4]])
    m.AddAllowedAssignments([t[3]],[[1],[2]])

    s=cp_model.CpSolver()
    if s.Solve(m)==4:
        for i in range(6):
            print(s.Value(t[i]),s.Value(r[i]))
```

```python
solve_course_timetabling()
```

```python
from ortools.sat.python import cp_model

def solve_vehicle_assignment():
    m=cp_model.CpModel()

    a=[m.NewIntVar(1,3,str(i)) for i in range(5)]

    x=[]
    for i in range(5):
        p=[]
        for j in range(1,4):
            q=m.NewBoolVar(str(i)+str(j))
            m.Add(a[i]==j).OnlyEnforceIf(q)
            m.Add(a[i]!=j).OnlyEnforceIf(q.Not())
            p.append(q)
        x.append(p)

    m.Add(sum([x[i][0] for i in range(5)])<=2)
    m.Add(sum([x[i][1] for i in range(5)])<=2)
    m.Add(sum([x[i][2] for i in range(5)])<=3)

    m.Add(a[0]!=a[1])
    m.Add(a[2]==3)

    c1=sum([x[i][0] for i in range(5)])
    c2=sum([x[i][1] for i in range(5)])

    d=m.NewIntVar(-5,5,'d')
    z=m.NewIntVar(0,5,'z')
    m.Add(d==c1-c2)
    m.AddAbsEquality(z,d)
    m.Add(z<=1)

    m.Add(a[4]==1).OnlyEnforceIf(x[3][0])

    s=cp_model.CpSolver()
    if s.Solve(m)==4:
        for i in range(5):
            print(s.Value(a[i]))
```

```python
solve_vehicle_assignment()
```

```python
from ortools.sat.python import cp_model

def solve_exam_schedule():
    m=cp_model.CpModel()

    a=[m.NewIntVar(1,6,str(i)) for i in range(5)]

    m.Add(a[0]!=a[1])
    m.Add(a[2]!=a[3])
    m.Add(a[3]!=a[4])

    x=m.NewIntVar(-5,5,'x')
    y=m.NewIntVar(2,6,'y')
    m.Add(x==a[2]-a[3])
    m.AddAbsEquality(y,x)

    m.Add(a[0]<a[4])

    for d in [[1,2],[3,4],[5,6]]:
        p=[]
        for i in range(5):
            q=m.NewBoolVar(str(i)+str(d[0]))
            m.AddAllowedAssignments([a[i]],[[d[0]],[d[1]]]).OnlyEnforceIf(q)
            m.AddForbiddenAssignments([a[i]],[[d[0]],[d[1]]]).OnlyEnforceIf(q.Not())
            p.append(q)
        m.Add(sum(p)<=2)

    s=cp_model.CpSolver()
    if s.Solve(m)==4:
        for i in range(5):
            print(s.Value(a[i]))
```

```python
solve_exam_schedule()
```
