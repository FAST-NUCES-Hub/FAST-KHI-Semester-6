Q1:

```python
import random

def make():
    a=[]
    b=[0,1,2,3,4,5,6]
    while len(b)>0:
        c=random.randint(0,len(b)-1)
        a.append(b[c])
        b.pop(c)
    return a

def fit(a):
    s=0
    i=0
    while i<7:
        j=i+1
        while j<7:
            if abs(i-j)!=abs(a[i]-a[j]):
                s=s+1
            j=j+1
        i=i+1
    return s

def pick(p,f):
    t=0
    for i in f:
        t=t+i
    r=random.randint(0,t)
    s=0
    i=0
    while i<len(p):
        s=s+f[i]
        if s>=r:
            return p[i]
        i=i+1

def cross(a,b):
    p=random.randint(1,5)
    c=a[:p]
    i=0
    while i<7:
        if b[i] not in c:
            c.append(b[i])
        i=i+1
    return c

def mut(a):
    i=random.randint(0,6)
    j=random.randint(0,6)
    t=a[i]
    a[i]=a[j]
    a[j]=t
    return a

def run():
    p=[]
    i=0
    while i<6:
        p.append(make())
        i=i+1

    g=0
    best=[]
    bestf=0

    while g<200:
        f=[]
        for i in p:
            f.append(fit(i))

        i=0
        while i<len(p):
            if f[i]>bestf:
                bestf=f[i]
                best=p[i]
            i=i+1

        if bestf==21:
            break

        np=[]
        i=0
        while i<6:
            a=pick(p,f)
            b=pick(p,f)
            c=cross(a,b)
            if random.random()<0.3:
                c=mut(c)
            np.append(c)
            i=i+1

        p=np
        g=g+1

    return best,bestf
```

```python
a,b=run()
print("Camera Placement:",a)
print("Fitness:",b)
```

```python
import random

def r(a,b):
    t=0
    for i in b:
        t=t+i
    x=random.randint(0,t)
    s=0
    i=0
    while i<len(a):
        s=s+b[i]
        if s>=x:
            return a[i]
        i=i+1

def t(a,b,k=3):
    c=[]
    i=0
    while i<k:
        x=random.randint(0,len(a)-1)
        c.append(x)
        i=i+1

    m=c[0]
    i=1
    while i<len(c):
        if b[c[i]]>b[m]:
            m=c[i]
        i=i+1

    return a[m]

def rk(a,b):
    p=list(zip(a,b))
    p.sort(key=lambda x:x[1])

    r=[]
    i=1
    while i<=len(a):
        r.append(i)
        i=i+1

    s=sum(r)
    x=random.randint(0,s)

    t1=0
    i=0
    while i<len(p):
        t1=t1+r[i]
        if t1>=x:
            return p[i][0]
        i=i+1
```

```python
a=['C1','C2','C3','C4','C5']
b=[12,20,15,8,25]

print("Roulette:",r(a,b))
print("Tournament:",t(a,b))
print("Rank:",rk(a,b))
```

```python
import random

def o(a,b):
    p=random.randint(1,5)
    c=a[:p]+b[p:]
    return c

def tw(a,b):
    p=random.randint(1,4)
    q=random.randint(p+1,6)
    c=a[:p]+b[p:q]+a[q:]
    return c

def ox(a,b):
    p=random.randint(1,4)
    q=random.randint(p+1,6)

    c=[-1]*7

    i=p
    while i<=q:
        c[i]=a[i]
        i=i+1

    k=0
    i=0
    while i<7:
        if b[i] not in c:
            while c[k]!=-1:
                k=k+1
            c[k]=b[i]
        i=i+1

    return c
```

```python
a=[1,3,5,0,6,4,2]
b=[2,5,1,6,0,3,4]

print("One Point:",o(a,b))
print("Two Point:",tw(a,b))
print("Order:",ox(a,b))
```

```python
import random

def s(a):
    i=random.randint(0,6)
    j=random.randint(0,6)
    t=a[i]
    a[i]=a[j]
    a[j]=t
    return a

def sc(a):
    i=random.randint(0,4)
    j=random.randint(i+1,6)

    b=a[i:j+1]

    k=len(b)-1
    while k>0:
        x=random.randint(0,k)
        t=b[k]
        b[k]=b[x]
        b[x]=t
        k=k-1

    a[i:j+1]=b
    return a

def inv(a):
    i=random.randint(0,4)
    j=random.randint(i+1,6)

    b=a[i:j+1]
    b=b[::-1]

    a[i:j+1]=b
    return a
```

```python
a=[1,3,5,0,6,4,2]

print("Swap:",s(a.copy()))
print("Scramble:",sc(a.copy()))
print("Inversion:",inv(a.copy()))
```
