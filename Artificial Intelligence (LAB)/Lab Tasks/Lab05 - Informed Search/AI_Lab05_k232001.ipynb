Q1:

```python
g = {
"A":[("B",4,2),("C",3,6)],
"B":[("D",5,1),("E",6,4)],
"C":[("F",4,3)],
"D":[("H",5,2)],
"E":[("H",3,1)],
"F":[("H",6,2)],
"H":[]
}

h = {"A":10,"B":8,"C":7,"D":5,"E":3,"F":6,"H":0}

def gb(s,t,lim):
    o=[(s,[s],0)]
    c=[]
    e=0
    while o:
        m=0
        for i in range(len(o)):
            if h[o[i][0]]<h[o[m][0]]:
                m=i
        n=o.pop(m)
        v,p,cost=n
        if v==t:
            return p,cost,e
        if v not in c:
            c.append(v)
            e+=1
            for x in g[v]:
                if x[2]<=lim:
                    o.append((x[0],p+[x[0]],cost+x[1]+x[2]))
    return [],0,e

def ast(s,t,lim):
    o=[(s,[s],0)]
    c=[]
    e=0
    while o:
        m=0
        for i in range(len(o)):
            if o[i][2]+h[o[i][0]] < o[m][2]+h[o[m][0]]:
                m=i
        n=o.pop(m)
        v,p,cost=n
        if v==t:
            return p,cost,e
        if v not in c:
            c.append(v)
            e+=1
            for x in g[v]:
                if x[2]<=lim:
                    o.append((x[0],p+[x[0]],cost+x[1]+x[2]))
    return [],0,e
```

```python
s="A"
t="H"
lim=4

p1,c1,e1=gb(s,t,lim)
print("GBFS Path:",p1)
print("Cost:",c1)
print("Expanded:",e1)

p2,c2,e2=ast(s,t,lim)
print("A* Path:",p2)
print("Cost:",c2)
print("Expanded:",e2)
```

Q2:

```python
g={
"A":[("B",4),("C",6)],
"B":[("D",5),("E",7)],
"C":[("F",3)],
"D":[("G",4)],
"E":[("G",2)],
"F":[("G",6)],
"G":[]
}

h={"A":9,"B":7,"C":6,"D":4,"E":3,"F":5,"G":0}

def gb(s,t):
    o=[(s,[s],0)]
    c=[]
    while o:
        m=0
        for i in range(len(o)):
            if h[o[i][0]]<h[o[m][0]]:
                m=i
        n=o.pop(m)
        v,p,cost=n
        if v==t:
            return p,cost
        if v not in c:
            c.append(v)
            for x in g[v]:
                o.append((x[0],p+[x[0]],cost+x[1]))
    return [],0

def ast(s,t):
    o=[(s,[s],0)]
    c=[]
    while o:
        m=0
        for i in range(len(o)):
            if o[i][2]+h[o[i][0]] < o[m][2]+h[o[m][0]]:
                m=i
        n=o.pop(m)
        v,p,cost=n
        if v==t:
            return p,cost
        if v not in c:
            c.append(v)
            for x in g[v]:
                o.append((x[0],p+[x[0]],cost+x[1]))
    return [],0

def learn(p,cost):
    for i in range(len(p)-1):
        h[p[i]]=cost
```

```python
s="A"
t="G"

p1,c1=gb(s,t)
print("Before learning:",p1,c1)

learn(p1,c1)

p2,c2=gb(s,t)
print("After learning:",p2,c2)

p3,c3=ast(s,t)
print("A*:",p3,c3)
```

```python
blk=[("B","E")]

s="A"
t="G"

p1,c1,e1=gb(s,t)
print("GBFS Path:",p1)
print("Time:",c1)
print("Expanded:",e1)

p2,c2,e2=ast(s,t)
print("A* Path:",p2)
print("Time:",c2)
print("Expanded:",e2)
```
