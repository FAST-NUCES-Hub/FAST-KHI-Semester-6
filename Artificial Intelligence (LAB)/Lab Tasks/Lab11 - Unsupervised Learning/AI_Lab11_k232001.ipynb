```python
from sklearn.preprocessing import LabelEncoder
a = LabelEncoder()
df['country'] = a.fit_transform(df['country'])
df.head()
```

```python
from sklearn.preprocessing import StandardScaler
```

```python
x = df.drop(['country'], axis=1)
scaler = StandardScaler()
x = pd.DataFrame(scaler.fit_transform(x))

print("Scaled data:")
x.head()
```

```python
df = df.dropna()
df.head()
```

```python
y = (df['income'] > df['income'].median()).astype(int)
x = df.drop(['country','income'], axis = 1)
```

```python
from sklearn.model_selection import train_test_split
x_train, x_test, y_train, y_test = train_test_split(x,y,test_size=0.2,random_state=42)
```

```python
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix

m1 = LogisticRegression(max_iter=5000)
m1.fit(x_train, y_train)
p1 = m1.predict(x_test)

print("Logistic Regression Accuracy:", accuracy_score(y_test, p1))
print("\nConfusion Matrix:\n", confusion_matrix(y_test, p1))
```

```python
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import accuracy_score, confusion_matrix

m2 = DecisionTreeClassifier(random_state=42)
m2.fit(x_train, y_train)
p2 = m2.predict(x_test)

print("Decision Tree Accuracy:", accuracy_score(y_test, p2))
print("\nConfusion Matrix:\n", confusion_matrix(y_test, p2))
```

```python
from sklearn.naive_bayes import GaussianNB
from sklearn.metrics import accuracy_score, confusion_matrix

m3 = GaussianNB()
m3.fit(x_train, y_train)
p3 = m3.predict(x_test)

print("Naive Bayes Accuracy:", accuracy_score(y_test, p3))
print("\nConfusion Matrix:\n", confusion_matrix(y_test, p3))
```

```python
from sklearn.cluster import KMeans

m4 = KMeans(n_clusters=3, random_state=42, n_init=10)
m4.fit(x)

p4 = m4.labels_
print("\nConfusion Matrix:\n", confusion_matrix(y, p4))

print("Cluster Labels:")
print(m4.labels_)
```
