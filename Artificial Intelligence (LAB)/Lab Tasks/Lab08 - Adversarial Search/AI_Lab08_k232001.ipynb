```python
def w(b):
    for i in range(3):
        if b[i][0]==b[i][1]==b[i][2]!=" ": return b[i][0]
        if b[0][i]==b[1][i]==b[2][i]!=" ": return b[0][i]
    if b[0][0]==b[1][1]==b[2][2]!=" ": return b[0][0]
    if b[0][2]==b[1][1]==b[2][0]!=" ": return b[0][2]
    return None

def f(b):
    for i in b:
        if " " in i: return False
    return True

def m(b,x):
    r=w(b)
    if r=="X": return 1
    if r=="O": return -1
    if f(b): return 0
    if x:
        v=-9
        for i in range(3):
            for j in range(3):
                if b[i][j]==" ":
                    b[i][j]="X"
                    v=max(v,m(b,0))
                    b[i][j]=" "
        return v
    else:
        v=9
        for i in range(3):
            for j in range(3):
                if b[i][j]==" ":
                    b[i][j]="O"
                    v=min(v,m(b,1))
                    b[i][j]=" "
        return v

def a(b):
    v=-9
    p=(0,0)
    for i in range(3):
        for j in range(3):
            if b[i][j]==" ":
                b[i][j]="X"
                t=m(b,0)
                b[i][j]=" "
                if t>v:
                    v=t
                    p=(i,j)
    return p

def p(b):
    for i in range(3):
        print(b[i][0],"|",b[i][1],"|",b[i][2])
    print()

def g():
    b=[[" "," "," "],[" "," "," "],[" "," "," "]]
    while True:
        p(b)
        r,c=map(int,input("Enter row and column (0-2): ").split())
        if b[r][c]!=" ":
            continue
        b[r][c]="O"
        if w(b) or f(b):
            p(b)
            break
        i,j=a(b)
        b[i][j]="X"
        if w(b) or f(b):
            p(b)
            break
```

```python
g()
```

```python
def f(a,l,r,x,al,be):
    if l>r: return 0
    if x:
        v=-9
        v=max(v,a[l]+f(a,l+1,r,0,al,be))
        al=max(al,v)
        if be<=al: return v
        v=max(v,a[r]+f(a,l,r-1,0,al,be))
        return v
    else:
        v=9
        v=min(v,f(a,l+1,r,1,al,be))
        be=min(be,v)
        if be<=al: return v
        v=min(v,f(a,l,r-1,1,al,be))
        return v
```

```python
a=[3,9,1,2,7,5]

l=0
r=len(a)-1
mx=0
mn=0
t=1

print("Initial Coins:",a)

while l<=r:
    if t:
        if f(a,l+1,r,0,-9,9)+a[l] >= f(a,l,r-1,0,-9,9)+a[r]:
            print("Max picks",a[l],", Remaining Coins:",a[l+1:r+1])
            mx+=a[l]
            l+=1
        else:
            print("Max picks",a[r],", Remaining Coins:",a[l:r])
            mx+=a[r]
            r-=1
    else:
        if a[l] <= a[r]:
            print("Min picks",a[l],", Remaining Coins:",a[l+1:r+1])
            mn+=a[l]
            l+=1
        else:
            print("Min picks",a[r],", Remaining Coins:",a[l:r])
            mn+=a[r]
            r-=1
    t=1-t

print("Final Scores - Max:",mx,", Min:",mn)
if mx>mn:
    print("Winner: Max")
else:
    print("Winner: Min")
```

```python
def v(b):
    s=0
    for i in b:
        for j in i:
            if j=="p": s+=1
            if j=="r": s+=5
            if j=="q": s+=9
            if j=="P": s-=1
            if j=="R": s-=5
            if j=="Q": s-=9
    return s

def f(b,d,x,al,be):
    if d==0:
        return v(b)
    if x:
        m=-9
        for i in range(8):
            for j in range(8):
                if b[i][j]!=" ":
                    m=max(m,v(b))
                    al=max(al,m)
                    if be<=al: return m
        return m
    else:
        m=9
        for i in range(8):
            for j in range(8):
                if b[i][j]!=" ":
                    m=min(m,v(b))
                    be=min(be,m)
                    if be<=al: return m
        return m
```

```python
b=[[" "]*8 for _ in range(8)]
b[0][0]="q"
b[7][7]="Q"

print("Initial Board:")
for i in b:
    print(i)

print()

for i in range(8):
    for j in range(8):
        if b[i][j]!=" ":
            print("Piece at position:",(i,j),"=",b[i][j],"-> evaluated value:",v(b))

print()
print("Best evaluation:",f(b,2,1,-9,9))
```

```python
def f(a,x):
    if not a: return 0
    if x:
        v=-9
        for i in range(len(a)):
            t=a[:i]+a[i+1:]
            v=max(v,1+f(t,0))
        return v
    else:
        v=9
        for i in range(len(a)):
            t=a[:i]+a[i+1:]
            v=min(v,f(t,1))
        return v
```

```python
a=["a","b","c","d"]

r=f(a,1)

print("Score:",r)

if r>0:
    print("Max can form longer sequence")
else:
    print("Min can reduce sequence")

print("One possible word:")
print("".join(a))
```
