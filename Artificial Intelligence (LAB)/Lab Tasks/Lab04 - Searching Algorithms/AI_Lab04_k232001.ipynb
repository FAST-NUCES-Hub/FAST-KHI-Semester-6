Q1: Goal Based Agent

```python
class A:

    def __init__(s,g,l):
        s.g=g
        s.l=l

    def f(s,n,t,d,p):
        if n==t:
            return p
        if d==0:
            return None
        for i in s.g[n]:
            r=s.f(i,t,d-1,p+[i])
            if r:
                return r
        return None
```

```python
g={
'A':['B','C'],
'B':['D','E'],
'C':['F'],
'D':[],
'E':[],
'F':[]
}

a=A(g,3)

print(a.f('A','E',3,['A']))
```

Q1: Utility Based Agent

```python
class A:

    def __init__(s,g):
        s.g=g

    def f(s,a,b):
        q=[[0,a,[a]]]
        v={}
        while q:
            q.sort()
            c,n,p=q.pop(0)
            if n==b:
                return p,c
            if n in v and v[n]<=c:
                continue
            v[n]=c
            for i in s.g[n]:
                q.append([c+s.g[n][i],i,p+[i]])
```

```python
g={
'A':{'B':1,'C':4},
'B':{'D':2,'E':5},
'C':{'F':1},
'D':{},
'E':{},
'F':{}
}

a=A(g)

print(a.f('A','F'))
```

```python
class A:

    def __init__(s,g,l):
        s.g=g
        s.l=l

    def h(s,x):
        return x==s.g

    def m(s,x):
        r=[]
        i=x.index(0)

        if i%3!=0:
            t=x[:]
            t[i],t[i-1]=t[i-1],t[i]
            r.append(t)

        if i%3!=2:
            t=x[:]
            t[i],t[i+1]=t[i+1],t[i]
            r.append(t)

        if i>2:
            t=x[:]
            t[i],t[i-3]=t[i-3],t[i]
            r.append(t)

        if i<6:
            t=x[:]
            t[i],t[i+3]=t[i+3],t[i]
            r.append(t)

        return r

    def f(s,x,d,p):

        if s.h(x):
            return p

        if d==0:
            return None

        for i in s.m(x):

            if i not in p:

                r=s.f(i,d-1,p+[i])

                if r:
                    return r
```

```python
g=[1,2,3,4,5,6,7,8,0]

a=A(g,5)

s=[1,2,3,4,5,6,7,0,8]

print(a.f(s,5,[s]))
```

```python
class A:

    def __init__(s,g):
        s.g=g

    def d(s,n,t,l,p):

        if n==t:
            return p

        if l==0:
            return None

        for i in s.g[n]:

            r=s.d(i,t,l-1,p+[i])

            if r:
                return r

    def f(s,a,b):

        for i in range(1,10):

            r=s.d(a,b,i,[a])

            if r:
                return r
```

```python
g={
'A':['B','C'],
'B':['D'],
'C':['E'],
'D':[],
'E':[]
}

a=A(g)

print(a.f('A','E'))
```

```python
g=[
[5,3,0,0,7,0,0,0,0],
[6,0,0,1,9,5,0,0,0],
[0,9,8,0,0,0,0,6,0],
[8,0,0,0,6,0,0,0,3],
[4,0,0,8,0,3,0,0,1],
[7,0,0,0,2,0,0,0,6],
[0,6,0,0,0,0,2,8,0],
[0,0,0,4,1,9,0,0,5],
[0,0,0,0,8,0,0,7,9]
]

def ok(r,c,x):
    for i in range(9):
        if g[r][i]==x:
            return False
        if g[i][c]==x:
            return False
    sr=r-r%3
    sc=c-c%3
    for i in range(3):
        for j in range(3):
            if g[sr+i][sc+j]==x:
                return False
    return True

def f():
    for r in range(9):
        for c in range(9):
            if g[r][c]==0:
                for x in range(1,10):
                    if ok(r,c,x):
                        g[r][c]=x
                        if f():
                            return True
                        g[r][c]=0
                return False
    return True
```
